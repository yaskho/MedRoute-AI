```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split

# 1. Dataset enrichi de requêtes médicales pour l'entraînement du routeur
data = {
    "text": [
        # TABULAR_ONLY (Diabète / Données structurées)
        "Est-ce que ce patient présente un risque de diabète selon ses données ?",
        "Analyser ces valeurs biologiques pour évaluer la glycémie et le profil métabolique.",
        "Vérifier les antécédents et les paramètres structurés du patient pour le diabète.",
        "Le patient a-t-il des indicateurs cliniques concordant avec le diabète ?",
        "Évaluer le taux de glucose, l'IMC et l'âge du patient.",
        "Est-ce que le profil tabulaire indique un diabète ?",
        "Données biologiques et taux de sucre dans le sang du patient.",
        "Analyser le dossier structuré du patient diabétique.",
        
        # IMAGE_ONLY (Mélanome / Peau)
        "Examiner cette photo de lésion cutanée pour voir si c'est un mélanome.",
        "Pouvez-vous analyser cette image de grain de beauté suspect ?",
        "Classifier cette image dermatologique pour exclure un carcinome ou un mélanome.",
        "Regarder la photo de cette tache sur la peau.",
        "Analyser l'image de la lésion cutanée fournie.",
        "Est-ce un mélanome malin sur cette photo ?",
        "Photo d'une lésion dermatologique suspecte.",
        
        # MULTIMODAL (Les deux)
        "Utiliser à la fois les données du patient et l'image de la lésion pour une analyse complète.",
        "Faire une évaluation globale combinant les paramètres structurés et l'image dermatologique.",
        "Je souhaite une analyse multimodale avec le dossier médical et la photo de la peau.",
        "Analyser l'ensemble des informations disponibles, à la fois tabulaires et visuelles.",
        "Croiser les données du diabète avec l'image de la lésion cutanée.",
        "Utiliser toutes les sources d'information pour ce cas."
    ],
    "label": [
        0, 0, 0, 0, 0, 0, 0, 0,  # TABULAR_ONLY
        1, 1, 1, 1, 1, 1, 1,     # IMAGE_ONLY
        2, 2, 2, 2, 2, 2         # MULTIMODAL
    ]
}

df_router = pd.DataFrame(data)
print(df_router['label'].value_counts())
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Paramètres de tokenisation
vocab_size = 1000
max_length = 20
oov_tok = "<OOV>"

# 2. Séparation des textes et des labels
training_sentences = df_router["text"].values
training_labels = df_router["label"].values

# 3. Initialisation et entraînement du Tokenizer
tokenizer = Tokenizer(num_words=vocab_size, oov_token=oov_tok)
tokenizer.fit_on_texts(training_sentences)

# 4. Transformation des phrases en séquences d'entiers
train_sequences = tokenizer.texts_to_sequences(training_sentences)
train_padded = pad_sequences(train_sequences, maxlen=max_length, padding='post', truncating='post')

print("Exemple de texte converti en séquence :")
print(train_padded[0])
```

```python
# 1. Construction du modèle Transformer simple pour la classification d'intentions
embedding_dim = 64
num_heads = 2
ff_dim = 64

inputs = layers.Input(shape=(max_length,))
embedding_layer = layers.Embedding(input_dim=vocab_size, output_dim=embedding_dim)(inputs)

# Couche d'attention (Bloc Transformer basique)
attention_output = layers.MultiHeadAttention(num_heads=num_heads, key_dim=embedding_dim)(embedding_layer, embedding_layer)
x = layers.LayerNormalization(epsilon=1e-6)(attention_output + embedding_layer)

# Couche Feed-Forward
ffn_output = tf.keras.Sequential([
    layers.Dense(ff_dim, activation="relu"),
    layers.Dense(embedding_dim),
])(x)
x = layers.LayerNormalization(epsilon=1e-6)(ffn_output + x)

# Pooling et classification finale
x = layers.GlobalAveragePooling1D()(x)
x = layers.Dropout(0.1)(x)
x = layers.Dense(32, activation="relu")(x)
outputs = layers.Dense(3, activation="softmax")(x)

transformer_model = tf.keras.Model(inputs=inputs, outputs=outputs)

# Compilation du modèle
transformer_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

transformer_model.summary()
```

```python
# 1. Entraînement du modèle Transformer (Routeur)
history = transformer_model.fit(
    train_padded, 
    training_labels, 
    epochs=35, 
    verbose=1
)

# 2. Test rapide d'une requête de routage
test_queries = [
    "Analyser la glycémie de ce patient",
    "Est-ce un mélanome sur cette photo ?",
    "Je veux utiliser toutes les informations disponibles"
]

# Tokenisation des tests
test_sequences = tokenizer.texts_to_sequences(test_queries)
test_padded = pad_sequences(test_sequences, maxlen=max_length, padding='post', truncating='post')

# Prédictions
predictions = transformer_model.predict(test_padded)
predicted_classes = np.argmax(predictions, axis=1)

intent_mapping = {0: "TABULAR_ONLY", 1: "IMAGE_ONLY", 2: "MULTIMODAL"}

print("\n--- Test du Routeur Intelligent ---")
for query, pred_idx in zip(test_queries, predicted_classes):
    print(f"Requête : '{query}' --> Route vers : {intent_mapping[pred_idx]}")
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.sequence import pad_sequences
import joblib

# 1. Chargement des modèles spécialisés et du routeur préalablement sauvegardés
rf_model = joblib.load("../models/diabetes_random_forest.pkl")
melanoma_model = tf.keras.models.load_model("medroute_melanoma_model.h5")
# (Note: Assure-toi d'avoir aussi sauvegardé ton tokenizer et ton transformer model si ce n'est pas déjà fait)

def medroute_orchestrator(prompt, tabular_features=None, image_path=None):
    """
    Orchestrateur central MedRoute AI :
    1. Analyse la requête texte via le routeur Transformer.
    2. Détermine le chemin (TABULAR_ONLY, IMAGE_ONLY, MULTIMODAL).
    3. Exécute le ou les modèles correspondants.
    """
    print(f"\n--- Requête reçue : '{prompt}' ---")
    
    # Étape A : Routage par le Transformer
    seq = tokenizer.texts_to_sequences([prompt])
    padded_seq = pad_sequences(seq, maxlen=max_length, padding='post', truncating='post')
    prediction = transformer_model.predict(padded_seq, verbose=0)
    intent_idx = np.argmax(prediction)
    
    intent_mapping = {0: "TABULAR_ONLY", 1: "IMAGE_ONLY", 2: "MULTIMODAL"}
    selected_path = intent_mapping[intent_idx]
    print(f"[Routeur AI] Chemin sélectionné --> {selected_path}")
    
    results = {}
    
    # Étape B : Exécution conditionnelle selon le chemin
    if selected_path == "TABULAR_ONLY" or selected_path == "MULTIMODAL":
        if tabular_features is not None:
            # Prédiction du modèle Tabulaire (Diabète)
            tab_pred = rf_model.predict([tabular_features])
            tab_prob = rf_model.predict_proba([tabular_features])
            results["tabular_prediction"] = int(tab_pred[0])
            results["tabular_probability"] = float(np.max(tab_prob))
            print(f"-> Modèle Tabulaire exécuté : Risque de diabète = {tab_pred[0]} (Confiance: {np.max(tab_prob)*100:.2f}%)")
        else:
            print("-> Avertissement : Données tabulaires requises mais non fournies.")
            
    if selected_path == "IMAGE_ONLY" or selected_path == "MULTIMODAL":
        if image_path is not None:
            # Appel de ta fonction de prédiction d'image (MobileNetV2 / Melanoma)
            # (Tu peux intégrer ici ta fonction predict_skin_lesion validée précédemment)
            print(f"-> Modèle d'imagerie exécuté sur : {image_path}")
            # results["image_prediction"] = ...
        else:
            print("-> Avertissement : Image requise mais non fournie.")
            
    return results

# Exemple de test global de la plateforme unifiée :
# medroute_orchestrator("Analyser la glycémie de ce patient", tabular_features=[...])
```

```python
import numpy as np
from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
import torch

# 1. Préparation des données (Exemple de structure)
# Supposons que tu as une liste de textes 'texts' et leurs labels correspondants 'labels' 
# (0: TABULAR_ONLY, 1: IMAGE_ONLY, 2: MULTIMODAL)
# X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.2, random_state=42)

# 2. Chargement du Tokenizer (ex: DistilBERT ou BERT miniature pour la médecine/rapidité)
MODEL_CHECKPOINT = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT)

def tokenize_function(examples):
    return tokenizer(examples["text"], padding="max_length", truncation=True, max_length=128)

# Conversion en Dataset Hugging Face (à adapter avec tes listes train/test)
# train_dataset = Dataset.from_dict({"text": X_train, "label": y_train}).map(tokenize_function, batched=True)
# test_dataset = Dataset.from_dict({"text": X_test, "label": y_test}).map(tokenize_function, batched=True)

# 3. Définition des métriques d'évaluation
def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    acc = accuracy_score(labels, preds)
    return {"accuracy": acc}

# 4. Chargement du modèle pour 3 classes de routage
model = AutoModelForSequenceClassification.from_pretrained(MODEL_CHECKPOINT, num_labels=3)

# 5. Configuration des hyperparamètres d'entraînement (Optimisation)
training_args = TrainingArguments(
    output_dir="../models/transformer/results",
    eval_strategy="epoch",            # Évaluer à chaque fin d'époque
    save_strategy="epoch",
    learning_rate=3e-5,               # Taux d'apprentissage optimisé pour le fine-tuning
    per_device_train_batch_size=16,   # Taille des lots
    per_device_eval_batch_size=16,
    num_train_epochs=3,               # Nombre d'époques
    weight_decay=0.01,
    load_best_model_at_end=True,
    metric_for_best_model="accuracy",
    logging_steps=10,
    seed=42
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    compute_metrics=compute_metrics,
)

# 6. Lancement du fine-tuning
print("Lancement du fine-tuning du routeur Transformer...")
# trainer.train()

# 7. Évaluation finale et sauvegarde
# eval_results = trainer.evaluate()
# print(f"Résultats d'évaluation : {eval_results}")

# Sauvegarde finale du modèle et du tokenizer dans le dossier models/
model.save_pretrained("../models/transformer/best_router_model")
tokenizer.save_pretrained("../models/transformer/best_router_tokenizer")
print("Modèle Transformer et Tokenizer sauvegardés avec succès dans models/transformer/ !")
```
